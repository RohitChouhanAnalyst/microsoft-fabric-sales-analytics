```python
    from pyspark.sql.functions import col, to_date

    bronze_df = spark.table("bronze_sales")

    silver_df = bronze_df \
        .dropDuplicates() \
        .dropna() \
        .withColumn("OrderDate", to_date(col("OrderDate"), "dd-MM-yyyy"))

    silver_df.write \
        .mode("overwrite") \
        .format("delta") \
        .saveAsTable("silver_sales")

    display(spark.table("silver_sales"))
```
