```python
from pyspark.sql.functions import sum

silver_df = spark.table("silver_sales")

gold_region_sales = silver_df.groupBy("Region").agg(
    sum("Revenue").alias("Total_Revenue"),
    sum("Quantity").alias("Total_Quantity")
)

gold_region_sales.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("gold_region_sales")
```

```python
from pyspark.sql.functions import sum

gold_product_sales = silver_df.groupBy("ProductCategory").agg(
    sum("Revenue").alias("Total_Revenue"),
    sum("Quantity").alias("Total_Quantity")
)

gold_product_sales.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("gold_product_sales")
```

```python
from pyspark.sql.functions import count

gold_status_summary = silver_df.groupBy("Status").agg(
    count("OrderID").alias("Total_Orders"),
    sum("Revenue").alias("Total_Revenue")
)

gold_status_summary.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("gold_status_summary")
```

```python
from pyspark.sql.functions import month, year, sum

gold_monthly_sales = silver_df \
    .withColumn("Year", year("OrderDate")) \
    .withColumn("Month", month("OrderDate")) \
    .groupBy("Year", "Month") \
    .agg(
        sum("Revenue").alias("Total_Revenue"),
        sum("Quantity").alias("Total_Quantity")
    )

gold_monthly_sales.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("gold_monthly_sales")
```

```python
from pyspark.sql.functions import sum

gold_customer_sales = silver_df.groupBy("CustomerName").agg(
    sum("Revenue").alias("Total_Revenue"),
    sum("Quantity").alias("Total_Quantity")
)

gold_customer_sales.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("gold_customer_sales")
```

```python
from pyspark.sql.functions import sum

gold_top_categories = silver_df.groupBy("ProductCategory") \
    .agg(sum("Revenue").alias("Revenue")) \
    .orderBy("Revenue", ascending=False)

gold_top_categories.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("gold_top_categories")
```

```python
from pyspark.sql.functions import sum, avg, max, min

gold_revenue_kpi = silver_df.agg(
    sum("Revenue").alias("Total_Revenue"),
    avg("Revenue").alias("Average_Revenue"),
    max("Revenue").alias("Highest_Order"),
    min("Revenue").alias("Lowest_Order")
)

gold_revenue_kpi.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("gold_revenue_kpi")
```

```python
from pyspark.sql.functions import sum

gold_region_product = silver_df.groupBy(
    "Region",
    "ProductCategory"
).agg(
    sum("Revenue").alias("Revenue"),
    sum("Quantity").alias("Quantity")
)

gold_region_product.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("gold_region_product")
```
