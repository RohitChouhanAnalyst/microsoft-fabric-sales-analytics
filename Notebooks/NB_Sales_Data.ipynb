```python
df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("Files/raw/sales_data.csv")
```

```python
df.write \
  .mode("overwrite") \
  .format("delta") \
  .saveAsTable("bronze_sales")

# Step 3: Read Bronze
bronze_df = spark.table("bronze_sales")
display(bronze_df)

bronze_df.groupBy("OrderID").count().filter("count > 1").show(20, False)
```
